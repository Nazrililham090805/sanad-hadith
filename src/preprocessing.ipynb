## Import Library

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import ast
import re
from collections import Counter
```

## Load Data

```python
# pd.set_option('display.max_colwidth', None)

df = pd.read_csv('../data/raw/sanadset.csv', header=None)
df.columns = ['Hadith', 'Book', 'Num_hadith', 'Matn', 'Sanad', 'Sanad_Length']
df = df.iloc[1:].reset_index(drop=True)

# Pastikan kolom Sanad ada
assert 'Sanad' in df.columns, "❌ Kolom 'Sanad' tidak ditemukan!"

print(f" Data loaded: {len(df)} baris")
print(f" Kolom: {df.columns.tolist()}")
print(f" Tipe data: {df.dtypes.to_dict()}")

print(f"\n Sample data:")
print(df.head(3))

df
```

## Menghapus No Sanad di Kolom Sanad

```python
print(f"📊 Before filtering: {len(df)} baris")
print(f"   Baris dengan 'No SANAD': {(df['Sanad'] == 'No SANAD').sum()}")

# Hapus baris dengan 'No SANAD'
df = df[df['Sanad'] != 'No SANAD'].copy()

print(f"\n✅ After filtering: {len(df)} baris")
print(f"   Data yang dihapus: {(df['Sanad'] == 'No SANAD').sum()} baris")

print(f"\n📊 Remaining data info:")
print(df.info())
```
