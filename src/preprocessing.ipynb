### Ubah Sanad dari String ke List

```python
# Ubah nilai string yang merepresentasikan list menjadi object list Python

def parse_list(value):
    if isinstance(value, str):
        try:
            return ast.literal_eval(value)
        except (ValueError, SyntaxError):
            return value
    return value

df['Sanad'] = df['Sanad'].apply(parse_list)

print(df['Sanad'].head())
print(type(df['Sanad'].iloc[0]))
```

### Menghilangkan Harakat

```python
def remove_harakat(text): ## Remove Arabic diacritics (harakat)

    # Regex karakter harakat Arab
    pattern = r'[\u064B-\u065F\u0670\u06D6-\u06ED]'
    # Jika data berupa list
    if isinstance(text, list):
        return [
            re.sub(pattern, '', str(t))
            for t in text]
    # Jika data berupa string
    elif isinstance(text, str):
        return re.sub(pattern, '', text)
    # Jika bukan string/list
    return text

print("Removing harakat from Sanad...")

df['Sanad_no_harakat'] = df['Sanad'].apply(remove_harakat)

print("Harakat removal completed")
print("\n Sample result:\n")
print("Sanad asli:")
print(df['Sanad'].iloc[0])
print("\nSanad tanpa harakat:")
print(df['Sanad_no_harakat'].iloc[0])
```

### Mengecek Variasi Penulisan Nama Perawi 'عائشة' dan ابن عباس (untuk mencari dan menghitung semua variasi penulisan nama perawi dalam dataset sanad hadis)

```python
# List nama perawi yang ingin dicari variasinya
nama_perawi = ['عائشة', 'ابن عباس']

for kata_cari in nama_perawi:
    # Pola regex huruf toleran harakat
    harakat_range = r'[\u064B-\u065F\u0670\u06D6-\u06ED]*'
    pattern = harakat_range.join(list(kata_cari))

    # Kumpulkan semua variasi ke satu list
    semua_variasi = []

    for row in df['Sanad_no_harakat']:
        try:
            sanad_list = ast.literal_eval(row) if isinstance(row, str) else row
        except:
            sanad_list = [row]

        for teks in sanad_list:
            matches = re.findall(pattern, teks)
            semua_variasi.extend(matches)

    # Hitung kemunculan setiap variasi
    counter = Counter(semua_variasi)

    # Ubah jadi DataFrame biar rapi
    df_hasil = pd.DataFrame(counter.items(), columns=['Variasi', 'Count']).sort_values(by='Count', ascending=False)

    print(f"Variasi untuk '{kata_cari}':")
    print(df_hasil)
    print("\n" + "="*50 + "\n")
```

### Normalisasi Alif dan Ya

```python
def normalisasi_alif_ya(lst):
    if isinstance(lst, list):
        hasil = []
        for kata in lst:
            # 1. Seragamkan variasi alif jadi alif biasa
            kata = re.sub(r'[إأآ]', 'ا', kata)
            # 2. Seragamkan variasi ya jadi ي
            kata = re.sub(r'[ىېے]', 'ي', kata)
            hasil.append(kata)
        return hasil
    return lst

df['Sanad_no_harakat'] = df['Sanad_no_harakat'].apply(normalisasi_alif_ya)

print("Normalisasi alif dan ya selesai")
print("Sample result:")
print(df['Sanad_no_harakat'].head(10))
```

### Validasi Sanad sebagai List dan Hitung Kemunculan Kata

```python
#jika beberapa baris berformat string seperti "['a','b']"
def ensure_list(cell):
    if isinstance(cell, list):
        return cell
    if isinstance(cell, str) and cell.strip().startswith('['):
        try:
            return ast.literal_eval(cell)
        except Exception:
            return cell
    return cell

df['Sanad_no_harakat'] = df['Sanad_no_harakat'].apply(ensure_list)

#fungsi hitung (aman untuk list/str/NaN)
def count_occurrences(series, term):
    def cnt_cell(cell):
        if isinstance(cell, list):
            return sum(item.count(term) for item in cell if isinstance(item, str))
        elif isinstance(cell, str):
            return cell.count(term)
        else:
            return 0
    return int(series.apply(cnt_cell).sum())

kata_target = ['عائشة', 'ابن عباس']

for kata in kata_target:

    jumlah = count_occurrences(
        df['Sanad_no_harakat'],
        kata
    )

    print(f"Jumlah kemunculan '{kata}': {jumlah}")
```

### Mengganti kata kekerabatan pada anggota keluarga 

```python
def ganti_kata_kekerabatan(lst):
    result = []
    prev_asli = None  # simpan nama tanpa harakat (karena semua tanpa harakat)

    for word in lst:
        word_no_harak = word

        if word_no_harak in [
            'ابيه','اباها','اباه','ابيهما','ابيا','ابيك','اباك',
            'ابان','ابوك','ابوه','ابانا','ابي','اب']:
            if prev_asli:
                result.append(f'ابو {prev_asli}')
            else:
                # prev_asli belum ada, jadi jangan diubah
                result.append(word)


        elif word_no_harak in ['جديه', 'جده', 'جدها']:
            if prev_asli:
                result.append(f'جد {prev_asli}')
            else:
                result.append(word)

        elif word_no_harak in ['ابائه', 'ابائهم']:
            if prev_asli:
                result.append(f'آباء {prev_asli}')
            else:
                result.append(word)

        elif word_no_harak in ['ابيها', 'أبيها', 'لأبيها']:
            if prev_asli:
                result.append(f'ابو {prev_asli}')
            else:
                result.append(word)

        elif word_no_harak in ['امه', 'أمه', 'امها', 'أمها']:
            if prev_asli:
                result.append(f'ام {prev_asli}')
            else:
                result.append(word)

        elif word_no_harak in ['جدته', 'جدتها']:
            if prev_asli:
                result.append(f'جدة {prev_asli}')
            else:
                result.append(word)

        else:
            result.append(word)
            prev_asli = word_no_harak

    return result


# Terapkan ke satu kolom
df['Sanad_no_harakat'] = df['Sanad_no_harakat'].apply(ganti_kata_kekerabatan)

print("Penggantian kata kekerabatan selesai")
print("Sample result:")
print(df['Sanad_no_harakat'].head(10))
```

### Menormalisasi nama perawi (Aba Abi Abu) menjadi Abu

```python
def normalize_names(lst):
    normalized = []
    for word in lst:
        # ganti semua variasi ابا / ابي / ابو → ابو
        word = re.sub(r'\b(?:ابا|ابي|ابو)\b', 'ابو', word)
        normalized.append(word)
    return normalized

df['Sanad_no_harakat'] = df['Sanad_no_harakat'].apply(normalize_names)

print("Normalisasi kata 'abu' selesai")
print("Sample result:")
print(df['Sanad_no_harakat'].head(10))
```

### hapus

```python
transmission_words = {
    'حدثنا', 'حدثني', 'أخبرنا', 'أخبرني',
    'عن', 'قال', 'سمعت', 'روى'
}

remove_symbols_re = re.compile(r'[،:؛()\-/]')

def clean_list(lst):
    if not isinstance(lst, list):
        return lst

    result = []
    for x in lst:
        if not isinstance(x, str):
            continue

        text = x.strip()
        text = remove_symbols_re.sub('', text)

        if not text:
            continue

        if text in transmission_words:
            continue

        result.append(text)
    return result

df['Sanad_no_harakat'] = df['Sanad_no_harakat'].apply(clean_list)
print('Jumlah total baris di df:', len(df))
```
