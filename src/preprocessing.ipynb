### Menghapus No Sanad di Kolom Sanad

```python
# Hapus baris 'No SANAD'
df = df[df['Sanad'] != 'No SANAD'].copy()

# Print jumlah data setelah filtering
print("Jumlah data setelah No SANAD dihapus:", len(df))
df.head(5)
```

### Ubah Sanad dari String ke List

```python
# Ubah nilai string yang merepresentasikan list menjadi object list Python

def parse_list(value):
    if isinstance(value, str):
        try:
            return ast.literal_eval(value)
        except (ValueError, SyntaxError):
            return value
    return value

df['Sanad'] = df['Sanad'].apply(parse_list)

print(df['Sanad'].head())
print(type(df['Sanad'].iloc[0]))
```

### Menghilangkan Harakat

```python
def remove_harakat(text): ## Remove Arabic diacritics (harakat)

    # Regex karakter harakat Arab
    pattern = r'[\u064B-\u065F\u0670\u06D6-\u06ED]'
    # Jika data berupa list
    if isinstance(text, list):
        return [
            re.sub(pattern, '', str(t))
            for t in text]
    # Jika data berupa string
    elif isinstance(text, str):
        return re.sub(pattern, '', text)
    # Jika bukan string/list
    return text

print("Removing harakat from Sanad...")

df['Sanad_no_harakat'] = df['Sanad'].apply(remove_harakat)

print("Harakat removal completed")
print("\n Sample result:\n")
print("Sanad asli:")
print(df['Sanad'].iloc[0])
print("\nSanad tanpa harakat:")
print(df['Sanad_no_harakat'].iloc[0])
```

### Mengecek Variasi Penulisan Nama Perawi 'عائشة' dan ابن عباس (untuk mencari dan menghitung semua variasi penulisan nama perawi dalam dataset sanad hadis)

```python
# List nama perawi yang ingin dicari variasinya
nama_perawi = ['عائشة', 'ابن عباس']

for kata_cari in nama_perawi:
    # Pola regex huruf toleran harakat
    harakat_range = r'[\u064B-\u065F\u0670\u06D6-\u06ED]*'
    pattern = harakat_range.join(list(kata_cari))

    # Kumpulkan semua variasi ke satu list
    semua_variasi = []

    for row in df['Sanad_no_harakat']:
        try:
            sanad_list = ast.literal_eval(row) if isinstance(row, str) else row
        except:
            sanad_list = [row]

        for teks in sanad_list:
            matches = re.findall(pattern, teks)
            semua_variasi.extend(matches)

    # Hitung kemunculan setiap variasi
    counter = Counter(semua_variasi)

    # Ubah jadi DataFrame biar rapi
    df_hasil = pd.DataFrame(counter.items(), columns=['Variasi', 'Count']).sort_values(by='Count', ascending=False)

    print(f"Variasi untuk '{kata_cari}':")
    print(df_hasil)
    print("\n" + "="*50 + "\n")
```

### Normalisasi Alif dan Ya

```python
def normalisasi_alif_ya(lst):
    if isinstance(lst, list):
        hasil = []
        for kata in lst:
           # 1. Seragamkan variasi alif jadi alif biasa
            kata = re.sub(r'[إأآ]', 'ا', kata)
            # 2. Seragamkan variasi ya jadi ي
            kata = re.sub(r'[ىېے]', 'ي', kata)

    
            hasil.append(kata)
        return hasil
    return lst

df['Sanad_no_harakat'] = df['Sanad_no_harakat'].apply(normalisasi_alif_ya)

print("Normalisasi alif dan ya selesai")
print("Sample result:")
print(df['Sanad_no_harakat'].head(10))
```

### Validasi Sanad sebagai List dan Hitung Kemunculan Kata

```python
#jika beberapa baris berformat string seperti "['a','b']"
def ensure_list(cell):
    if isinstance(cell, list):
        return cell
    if isinstance(cell, str) and cell.strip().startswith('['):
        try:
            return ast.literal_eval(cell)
        except Exception:
            return cell
    return cell

df['Sanad_no_harakat'] = df['Sanad_no_harakat'].apply(ensure_list)

#fungsi hitung (aman untuk list/str/NaN)
def count_occurrences(series, term):
    def cnt_cell(cell):
        if isinstance(cell, list):
            return sum(item.count(term) for item in cell if isinstance(item, str))
        elif isinstance(cell, str):
            return cell.count(term)
        else:
            return 0
    return int(series.apply(cnt_cell).sum())

kata_target = ['عائشة', 'ابن عباس']

for kata in kata_target:

    jumlah = count_occurrences(
        df['Sanad_no_harakat'],
        kata
    )

    print(f"Jumlah kemunculan '{kata}': {jumlah}")
```

### Mengganti kata kekerabatan pada anggota keluarga 

```python
def ganti_kata_kekerabatan(lst):
    result = []
    prev_asli = None  # simpan nama tanpa harakat (karena semua tanpa harakat)

    for word in lst:
        word_no_harak = word

        if word_no_harak in [
            'ابيه','اباها','اباه','ابيهما','ابيا','ابيك','اباك',
            'ابان','ابوك','ابوه','ابانا','ابي','اب']:
            if prev_asli:
                result.append(f'ابو {prev_asli}')
            else:
                # prev_asli belum ada, jadi jangan diubah
                result.append(word)


        elif word_no_harak in ['جديه', 'جده', 'جدها']:
            if prev_asli:
                result.append(f'جد {prev_asli}')
            else:
                result.append(word)

        elif word_no_harak in ['ابائه', 'ابائهم']:
            if prev_asli:
                result.append(f'آباء {prev_asli}')
            else:
                result.append(word)

        elif word_no_harak in ['ابيها', 'أبيها', 'لأبيها']:
            if prev_asli:
                result.append(f'ابو {prev_asli}')
            else:
                result.append(word)

        elif word_no_harak in ['امه', 'أمه', 'امها', 'أمها']:
            if prev_asli:
                result.append(f'ام {prev_asli}')
            else:
                result.append(word)

        elif word_no_harak in ['جدته', 'جدتها']:
            if prev_asli:
                result.append(f'جدة {prev_asli}')
            else:
                result.append(word)

        else:
            result.append(word)
            prev_asli = word_no_harak

    return result


# Terapkan ke satu kolom
df['Sanad_no_harakat'] = df['Sanad_no_harakat'].apply(ganti_kata_kekerabatan)

print("Penggantian kata kekerabatan selesai")
print("Sample result:")
print(df['Sanad_no_harakat'].head(10))
```

### Menormalisasi nama perawi (Aba Abi Abu) menjadi Abu

```python
def normalize_names(lst):
    normalized = []
    for word in lst:
        # ganti semua variasi ابا / ابي / ابو → ابو
        word = re.sub(r'\b(?:ابا|ابي|ابو)\b', 'ابو', word)
        normalized.append(word)
    return normalized

df['Sanad_no_harakat'] = df['Sanad_no_harakat'].apply(normalize_names)

print("Normalisasi kata 'abu' selesai")
print("Sample result:")
print(df['Sanad_no_harakat'].head(10))
```

### Filter Baris yang Dimulai dengan Kata Kekerabatan

```python
target_words = [
    # أب variations 
    'ابيه','أبيه',
    'أباها','أباه',
    'ابيهما','أبيهما',
    'ابيا','أبيا',
    'ابيك','أبيك',
    'اباك','أباك',
    'ابان','أبان',
    'ابوك','أبوك',
    'ابوه','أبوه',
    'ابانا','أبانا',
    'ابي','ابيها','أبيها','لأبيها',
    'ابي', 'اب', 'اباه', 'اباها'

    # جد variations
    'جديه','جده','جدها','جدته','جدتها',

    # آباء variations
    'آبائه', 'ابائه', 'ابائهم'

    # أم variations
    'امه','أمه','امها','أمها'
]

def starts_with_target(lst):
    # pastikan bukan NaN dan berupa list
    if isinstance(lst, list) and len(lst) > 0:
        return lst[0] in target_words
    return False

# filter → ambil hanya yang BUKAN diawali target
df = df[~df['Sanad_no_harakat'].apply(starts_with_target)]

print("Filter selesai")
print(f"Jumlah baris setelah filter: {len(df)}")
print("Sample df setelah filter:")
print(df['Sanad_no_harakat'].head(5))
```

```python
def contains_exact_only(lst):
    target_words = [
    # أب variations
    'ابيه','أبيه',
    'أباها','أباه',
    'ابيهما','أبيهما',
    'ابيا','أبيا',
    'ابيك','أبيك',
    'اباك','أباك',
    'ابان','أبان',
    'ابوك','أبوك',
    'ابوه','أبوه',
    'ابانا','أبانا',
    'ابي','ابيها','أبيها','لأبيها',
    'ابي', 'اب', 'اباه', 'اباها'

    # جد variations
    'جديه','جده','جدها','جدته','جدتها',

    # آباء variations
    'آبائه', 'ابائه', 'ابائهم'

    # أم variations
    'امه','أمه','امها','أمها'
    ]# cek apakah list cuma berisi satu elemen dan sama dengan target
    return len(lst) == 1 and lst[0] in target_words

filtered_rows = df[df['Sanad_no_harakat'].apply(contains_exact_only)]

print("Filter selesai")
print(f"Jumlah baris yang hanya berisi satu kata kekerabatan: {len(filtered_rows)}")
print("Sample filtered_rows:")
print(filtered_rows['Sanad_no_harakat'].head(5))

filtered_rows
```

```python
transmission_words = {
    'حدثنا', 'حدثني', 'أخبرنا', 'أخبرني',
    'عن', 'قال', 'سمعت', 'روى'
}

remove_symbols_re = re.compile(r'[،:؛()\-/]')

def clean_list(lst):
    if not isinstance(lst, list):
        return lst

    result = []
    for x in lst:
        if not isinstance(x, str):
            continue

        text = x.strip()
        text = remove_symbols_re.sub('', text)

        if not text:
            continue

        if text in transmission_words:
            continue

        result.append(text)
    return result

df['Sanad_no_harakat'] = df['Sanad_no_harakat'].apply(clean_list)
print('Jumlah total baris di df:', len(df))
```

```python
# Daftar lafal periwayatan yang dihapus
lafal_periwayatan = [
    # Bentuk dasar
    'حدثنا', 'حدثني', 'حدث', 'حدثت',
    'اخبرنا', 'اخبرني', 'اخبر',
    'أخبرنا', 'أخبرني', 'أخبر',

    # Anba'
    'أنبأنا', 'انبانا',
    'أنبأني', 'انبأني',
    'انباني',

    # Qala dan turunannya
    'قال', 'قلت', 'يقول',
    'فقال', 'وقال', 'قالت',

    # Sami'a
    'سمعت', 'يسمع',

    # Riwayat
    'روى', 'يروي',

    # Kata penghubung sanad
    'عن', 'ذكر',

    # Singkatan sanad
    'ثنا', 'نا',

    # Tambahan umum sanad
    'قرأت', 'قرأنا',
    'كتب', 'كتب إلي',
    'أملاء', 'إملاء',
    'سألت', 'سألنا',
    'حدثكم', 'حدثهم',
    'اخبركم', 'أخبركم',
    'أن', 'إن'
]

noise_patterns = [
    r'اهل الكوفة',
    r'اهل البصرة',
    r'اهل المدينة',
    r'اهل مكة',
    r'\bالي\b'
]

noise_words = [
    'IDF',
    'يعني'
]


# Gabungkan lafal menjadi regex
lafal_pattern = r'\b(?:' + '|'.join(map(re.escape, lafal_periwayatan)) + r')\b'
noise_pattern = '|'.join(noise_patterns)

# Karakter/simbol yang dihapus
pattern = r'[,،#()="{}\-\?\.:\|]+|<IDF>|</IDF>|\bIDF\b'

# Gabungkan regex
noise_words = '|'.join(noise_words)


# Fungsi cleaning
def clean_text(text):

    if not isinstance(text, str):
        return text

    # Hapus lafal periwayatan
    text = re.sub(lafal_pattern, ' ', text)

    # Hapus noise tambahan
    text = re.sub(noise_pattern, ' ', text)

    # Hapus noise
    text = re.sub(noise_pattern, ' ', text)

    # Hapus tag IDF
    text = re.sub(r'<\/?IDF>', ' ', text)

    # Hapus harakat Arab
    text = re.sub(r'[ً-ٰٟ]', '', text)

    # Hapus tanda baca Arab dan simbol
    text = re.sub(r'[^\w\s؀-ۿ]', ' ', text)

    # Hapus waw di awal kata
    text = re.sub(r'^و(?=[؀-ۿ])', '', text)

    # Hapus tatweel
    text = re.sub(r'ـ+', '', text)

    # Hapus karakter/simbol khusus
    text = re.sub(pattern, '', text)

    # Hapus spasi berlebih
    text = re.sub(r'\s+', ' ', text).strip()

    # Hapus angka
    text = re.sub(r'\d+', '', text)

    # Hapus underscore
    text = text.replace('_', ' ')

    # hapus huruf ب di awal jika diikuti ابن
    text = re.sub(r'^ب(?=ابن)', '', text)


    return text

def ensure_list(cell):

    if isinstance(cell, list):
        return cell

    if isinstance(cell, str):

        # jika format string list
        if cell.strip().startswith('['):
            try:
                return ast.literal_eval(cell)
            except Exception:
                return [cell]

        return [cell]

    if pd.isna(cell):
        return []

    return [str(cell)]

# terapkan cleaning ke Sanad_no_harakat
df['Sanad_no_harakat'] = df['Sanad_no_harakat'].apply(ensure_list)

df['Sanad_no_harakat'] = df['Sanad_no_harakat'].apply(
    lambda sanad: [
        clean_text(x)
        for x in sanad
        if clean_text(x) != ''
    ]
)



# Tampilkan hasil
display(
    df[
        [
            'Sanad_no_harakat',
        ]
    ]
)

# Statistik jumlah sanad

print("Jumlah seluruh sanad :", len(df))
```

```python
# Daftar lafaz periwayatan
hapus_kata = [
    "ثنا",
    "حدثنا",
    "قال حدثنا",
    "وقال",
    "عن",
    "أخبرنا",
    "أنبأنا",
    "حدثني",
    "حدثه",
    "سمعت",
    "يقول",
    "ذكر",
    "رواه",
    "حدث",
    "حكى"
]

hapus_kata = sorted(hapus_kata, key=len, reverse=True)
pola = r'\b(?:' + '|'.join(map(re.escape, hapus_kata)) + r')\b'


# Pastikan data berupa list
def ensure_list(cell):

    # Jika sudah list
    if isinstance(cell, list):
        return cell

    # Jika string
    if isinstance(cell, str):

        # Jika format string list
        if cell.strip().startswith('['):
            try:
                return ast.literal_eval(cell)
            except:
                return [cell]

        return [cell]

    # Jika NaN
    if pd.isna(cell):
        return []

    return [str(cell)]

# Terapkan ke kolom sanad
df['Sanad_no_harakat'] = (df['Sanad_no_harakat'].apply(ensure_list))

# Fungsi mencari lafaz yang ditemukan
def cari_lafaz(sanad):

    ditemukan = []

    for item in sanad:
        hasil = re.findall(
            pola,
            str(item)
        )
        if hasil:
            ditemukan.extend(hasil)

    # Hilangkan duplikat
    return list(set(ditemukan))


# Ambil data yang mengandung lafaz
mengandung_lafaz = df[
    df['Sanad_no_harakat'].apply(
        lambda sanad: any(
            re.search(pola, str(x))
            for x in sanad
        )
    )
].copy()


# Tambahkan kolom lafaz ditemukan
mengandung_lafaz['Lafaz_Ditemukan'] = (
    mengandung_lafaz['Sanad_no_harakat']
    .apply(cari_lafaz)
)

# Tampilkan sebelum cleaning
print("DATA YANG MENGANDUNG LAFAZ PERIWAYATAN")


display(mengandung_lafaz[['Sanad_no_harakat','Lafaz_Ditemukan']])

# Fungsi membersihkan sanad
def clean_sanad_list(sanad):

    hasil = []
    for item in sanad:
        # Hapus lafaz
        item = re.sub(
            pola,
            '',
            str(item)
        )
        # Rapikan spasi
        item = re.sub(
            r'\s+',
            ' ',
            item
        ).strip()
        # Jangan masukkan kosong
        if item != '':
            hasil.append(item)

    return hasil

# Simpan hasil cleaning
# Terapkan ke kolom Sanad_no_harakat
df['Sanad_no_harakat'] = (df['Sanad_no_harakat'].apply(clean_sanad_list))

df = df.reset_index(drop=True)

# Hitung jumlah sanad yang mengandung lafaz
jumlah_mengandung = len(mengandung_lafaz)

print("\nJumlah sanad yang mengandung lafaz:",
      jumlah_mengandung)

df[['Sanad_no_harakat']].head()
```

```python
# Cek karakter selain huruf Arab
def check_non_arabic(text):

    if not isinstance(text, str):
        return False

    # Cari karakter selain Arab dan spasi
    return bool(
        re.search(r'[^\u0600-\u06FF\s]', text)
    )

# Ambil data yang masih memiliki
# karakter non-Arab pada kolom Guru atau Murid
non_arabic_df = df[
    df['Sanad_no_harakat'].apply(check_non_arabic)
]

# Tampilkan hasil
display(non_arabic_df[['Sanad_no_harakat']].head(20))

# Jumlah data
print("Jumlah data dengan karakter non-Arab:")
print(len(non_arabic_df))
```

```python
from collections import Counter

# Fungsi normalisasi nama Arab
def normalize_arabic_name(text):
    text = str(text)

    # Hilangkan spasi berlebih
    text = re.sub(r'\s+', ' ', text).strip()

    # Hapus akhiran " L" jika ada
    text = re.sub(r'\s+L$', '', text)

    # Normalisasi huruf Arab
    text = text.replace('أ', 'ا')
    text = text.replace('إ', 'ا')
    text = text.replace('آ', 'ا')

    text = text.replace('ى', 'ي')
    text = text.replace('ة', 'ه')

    return text

# Flatten semua nama perawi dari list Sanad_no_harakat
all_narrators = []

for sanad_list in df['Sanad_no_harakat']:
    if isinstance(sanad_list, list):
        all_narrators.extend(sanad_list)

# Normalisasi nama perawi
all_narrators_normalized = [
    normalize_arabic_name(name)
    for name in all_narrators
]

# Hitung frekuensi setelah normalisasi
narrator_counts = Counter(all_narrators_normalized)

# DataFrame hasil
narrator_df = pd.DataFrame(
    narrator_counts.most_common(),
    columns=['Perawi', 'Jumlah']
)

display(narrator_df)
```

```python
mapping = []

for sanad_list in df['Sanad_no_harakat']:
    if isinstance(sanad_list, list):
        for name in sanad_list:
            mapping.append({
                'Nama_Asli': name,
                'Nama_Normalisasi': normalize_arabic_name(name)
            })

mapping_df = pd.DataFrame(mapping)

merged = (
    mapping_df
    .groupby('Nama_Normalisasi')['Nama_Asli']
    .unique()
    .reset_index()
)

merged['Jumlah_Varian'] = merged['Nama_Asli'].apply(len)

merged = merged[
    merged['Jumlah_Varian'] > 1
].sort_values(
    'Jumlah_Varian',
    ascending=False
)

display(merged)

# Terapkan normalisasi kembali ke kolom Sanad_no_harakat

def normalize_sanad_list(sanad_list):
    if isinstance(sanad_list, list):
        return [normalize_arabic_name(name) for name in sanad_list]
    return sanad_list

# Perbarui kolom dengan daftar nama yang dinormalisasi

df['Sanad_no_harakat'] = df['Sanad_no_harakat'].apply(normalize_sanad_list)

print('Kolom Sanad_no_harakat telah dinormalisasi.')
df[['Sanad_no_harakat']].head()
```
