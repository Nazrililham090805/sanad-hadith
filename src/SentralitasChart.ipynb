# Visualisasi Chart Metrik Sentralitas

Notebook ini memuat data dari `Sentralitas.csv` dan membentuk grafik (chart) untuk setiap metrik sentralitas:
- In-Degree Centrality
- Out-Degree Centrality
- Eigenvector Centrality
- Closeness Centrality
- Betweenness Centrality

## Import Library

```python
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams['figure.dpi'] = 110
plt.rcParams['font.size'] = 10
```

## Load Data

```python
df = pd.read_csv('../data/processed/Sentralitas.csv')

print(f'Total perawi : {len(df):,}')
print(f'Kolom        : {list(df.columns)}')
df.head()
```

## Fungsi Pembuat Chart

Fungsi `plot_centrality` menampilkan Top-N perawi berdasarkan satu metrik sentralitas dalam bentuk bar chart vertikal. Nama perawi (`Nama_Perawi`, transliterasi) berada di sumbu X.

```python
TOP_N = 10

def plot_centrality(df, column, title, color):
    # Ambil Top-N berdasarkan nilai metrik (urut menurun)
    top = df.nlargest(TOP_N, column)

    fig, ax = plt.subplots(figsize=(10, 6))
    bars = ax.bar(top['Nama_Perawi'], top[column], color=color, edgecolor='black', linewidth=0.5)

    ax.set_title(title, fontsize=13, fontweight='bold')
    ax.set_xlabel('Perawi')
    ax.set_ylabel('Nilai Sentralitas')
    ax.set_xticklabels(top['Nama_Perawi'], rotation=45, ha='right')

    # Tampilkan nilai di atas tiap bar
    for bar in bars:
        h = bar.get_height()
        ax.text(bar.get_x() + bar.get_width() / 2, h,
                f'{h:.4f}', ha='center', va='bottom', fontsize=8)

    ax.margins(y=0.12)
    plt.tight_layout()
    plt.show()
```

## In-Degree Centrality
Perawi dengan jumlah murid terbanyak (otoritas sebagai sumber riwayat / guru).

```python
plot_centrality(df, 'In_Degree_Centrality',
                'Top 10 Perawi - In-Degree Centrality', '#4C72B0')
```

## Out-Degree Centrality
Perawi dengan jumlah guru terbanyak (keaktifan sebagai penerima riwayat / murid).

```python
plot_centrality(df, 'Out_Degree_Centrality',
                'Top 10 Perawi - Out-Degree Centrality', '#DD8452')
```

## Eigenvector Centrality
Perawi yang terhubung dengan perawi-perawi berpengaruh (prestise dalam jaringan).

```python
plot_centrality(df, 'Eigenvector_Centrality',
                'Top 10 Perawi - Eigenvector Centrality', '#C44E52')
```

## Closeness Centrality
Perawi dengan jalur periwayatan terpendek ke perawi lain (posisi sentral).

```python
plot_centrality(df, 'Closeness_Centrality',
                'Top 10 Perawi - Closeness Centrality', '#8172B3')
```

## Betweenness Centrality
Perawi yang menjadi jembatan antar-kelompok (peran penghubung dalam rantai sanad).

```python
plot_centrality(df, 'Betweenness_Centrality',
                'Top 10 Perawi - Betweenness Centrality', '#937860')
```

## Gabungan Semua Metrik (5 Subplot)
Menampilkan kelima metrik sentralitas dalam satu figur untuk perbandingan.

```python
metrics = [
    ('In_Degree_Centrality',   'In-Degree',    '#4C72B0'),
    ('Out_Degree_Centrality',  'Out-Degree',   '#DD8452'),
    ('Eigenvector_Centrality', 'Eigenvector',  '#C44E52'),
    ('Closeness_Centrality',   'Closeness',    '#8172B3'),
    ('Betweenness_Centrality', 'Betweenness',  '#937860'),
]

fig, axes = plt.subplots(2, 3, figsize=(20, 11))
axes = axes.flatten()

for ax, (col, label, color) in zip(axes, metrics):
    top = df.nlargest(TOP_N, col)
    ax.bar(top['Nama_Perawi'], top[col], color=color, edgecolor='black', linewidth=0.5)
    ax.set_title(f'Top 10 - {label} Centrality', fontsize=11, fontweight='bold')
    ax.set_xticklabels(top['Nama_Perawi'], rotation=45, ha='right', fontsize=8)
    ax.margins(y=0.12)

# Sembunyikan subplot ke-6 yang tidak terpakai
axes[5].axis('off')

plt.suptitle('Perbandingan Metrik Sentralitas Perawi Hadis', fontsize=15, fontweight='bold')
plt.tight_layout(rect=[0, 0, 1, 0.97])
plt.show()
```
