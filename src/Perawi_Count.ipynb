# Menghitung Count Perawi sebagai Guru dan Murid

## Import Library

```python
import pandas as pd
import re
```

```python
# pd.set_option('display.max_colwidth', None)

df = pd.read_csv('../data/processed/edges.csv', header=None)
df.columns = ['Guru', 'Murid', 'Weight', 'Inverse_Weight']
df = df.iloc[1:].reset_index(drop=True)
```

```python
display(df)
```

```python
edges_df = df[['Guru', 'Murid']].copy()

# Ambil semua nama perawi dari Guru dan Murid
semua_perawi = pd.concat([
    edges_df['Guru'],
    edges_df['Murid']
]).drop_duplicates().reset_index(drop=True)

# Buat DataFrame perawi
perawi_stats = pd.DataFrame({
    'Perawi': semua_perawi
})

# Hitung jumlah sebagai Guru
perawi_stats['Jumlah_Sebagai_Guru'] = (
    perawi_stats['Perawi']
    .map(edges_df['Guru'].value_counts())
    .fillna(0)
    .astype(int)
)

# Hitung jumlah sebagai Murid
perawi_stats['Jumlah_Sebagai_Murid'] = (
    perawi_stats['Perawi']
    .map(edges_df['Murid'].value_counts())
    .fillna(0)
    .astype(int)
)

# Tampilkan hasil
display(perawi_stats)
```
