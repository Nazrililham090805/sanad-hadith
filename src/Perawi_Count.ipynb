```python
# Hitung metrik perawi dari edges.csv dan sanadset_cleaned.csv
import pandas as pd
import re
import ast


def normalize_arabic_name(text):
    text = str(text)
    text = text.strip()
    text = re.sub(r"\s+", " ", text)
    return text


# Load edge relasi guru-murid
edges_df = pd.read_csv('../data/processed/edges.csv')
edges_df = edges_df[['Guru', 'Murid', 'Weight']].copy()
edges_df['Guru'] = edges_df['Guru'].astype(str).apply(normalize_arabic_name)
edges_df['Murid'] = edges_df['Murid'].astype(str).apply(normalize_arabic_name)
edges_df['Weight'] = pd.to_numeric(edges_df['Weight'], errors='coerce').fillna(1).astype(int)

# Load data hadis untuk menghitung jumlah hadis per perawi
hadith_df = pd.read_csv('../data/processed/sanadset_cleaned.csv')
hadith_df['Sanad_no_harakat'] = hadith_df['Sanad_no_harakat'].astype(str).str.strip()
hadith_df['Sanad_no_harakat'] = hadith_df['Sanad_no_harakat'].apply(lambda x: ast.literal_eval(x) if isinstance(x, str) and x.startswith('[') else x)

sanad_exploded = hadith_df.explode('Sanad_no_harakat')
sanad_exploded['Sanad_no_harakat'] = sanad_exploded['Sanad_no_harakat'].astype(str).apply(normalize_arabic_name)

# Ke semua perawi unik dari relasi edges
semua_perawi = pd.concat([edges_df['Guru'], edges_df['Murid']]).drop_duplicates().reset_index(drop=True)
perawi_stats = pd.DataFrame({'Perawi': semua_perawi})

# Jumlah guru unik yang dimiliki perawi
jumlah_guru = edges_df.groupby('Murid')['Guru'].nunique()
perawi_stats['Jumlah_Guru'] = perawi_stats['Perawi'].map(jumlah_guru).fillna(0).astype(int)

# Jumlah murid unik yang dimiliki perawi
jumlah_murid = edges_df.groupby('Guru')['Murid'].nunique()
perawi_stats['Jumlah_Murid'] = perawi_stats['Perawi'].map(jumlah_murid).fillna(0).astype(int)

# Total riwayat dari guru ke perawi (berdasarkan bobot relasi)
total_sebagai_murid = edges_df.groupby('Murid')['Weight'].sum()
perawi_stats['Total_Riwayat_dari_Guru'] = perawi_stats['Perawi'].map(total_sebagai_murid).fillna(0).astype(int)

# Total riwayat yang diteruskan perawi ke murid (berdasarkan bobot relasi)
total_sebagai_guru = edges_df.groupby('Guru')['Weight'].sum()
perawi_stats['Total_Riwayat_ke_Murid'] = perawi_stats['Perawi'].map(total_sebagai_guru).fillna(0).astype(int)

# Total relasi adalah jumlah guru + jumlah murid
perawi_stats['Total_Relasi'] = perawi_stats['Jumlah_Guru'] + perawi_stats['Jumlah_Murid']

# Jumlah hadis unik yang menyebut perawi dalam sanad
jumlah_hadis_perawi = sanad_exploded.groupby('Sanad_no_harakat')['Matn'].nunique()
perawi_stats['Jumlah_Hadis'] = perawi_stats['Perawi'].map(jumlah_hadis_perawi).fillna(0).astype(int)

# Tampilkan ringkasan perawi
perawi_stats = perawi_stats.sort_values(
    by=['Jumlah_Hadis', 'Total_Relasi', 'Jumlah_Guru', 'Jumlah_Murid'],
    ascending=False
).reset_index(drop=True)

display(perawi_stats.head(50))
```

```python
cek_sanad = hadith_df.copy()

cek_sanad['Sanad_String'] = (
    cek_sanad['Sanad_no_harakat']
    .astype(str)
)

cek_sanad = (
    cek_sanad
    .groupby('Matn')['Sanad_String']
    .nunique()
    .reset_index(name='Jumlah_Sanad')
)

cek_sanad_berbeda = (
    cek_sanad[
        cek_sanad['Jumlah_Sanad'] > 1
    ]
)

print("Matan dengan sanad berbeda:")
print(len(cek_sanad_berbeda))

display(
    cek_sanad_berbeda.sort_values(
        'Jumlah_Sanad',
        ascending=False
    )
)
```
